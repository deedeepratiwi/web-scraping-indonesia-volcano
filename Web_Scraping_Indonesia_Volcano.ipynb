```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
wiki_url = 'https://en.wikipedia.org/wiki/List_of_volcanoes_in_Indonesia'

response = requests.get(wiki_url)
```

```python
content = response.content
```

```python
parser = BeautifulSoup(content, 'html.parser')
```

```python
#print(parser.prettify())
```

```python
table_name = 'wikitable sortable' 
tables = parser.find_all('table', {'class':table_name})

len(tables)
```

```python
# We found seven tables, just like the website. The first six has almost similar coloumns and they are divided based on location.
# We are going to add one more column to note the location

name = []
shape = []
elevation = []
last_eruption = []
coordinates = []
island = []

rows = tables[0].find_all('tr')    
for row in rows[1:]:
    cells = row.find_all('td')
                
    name.append(cells[0].text.strip())
    shape.append(cells[1].text.strip())
    elevation.append(cells[2].text.strip())
    last_eruption.append(cells[3].text.strip())
    coordinates.append(cells[4].text.strip().split("/")[0])  
    island.append('Sumatra')

rows = tables[1].find_all('tr')    
for row in rows[1:]:
    cells = row.find_all('td')
                
    name.append(cells[0].text.strip())
    shape.append(cells[1].text.strip())
    elevation.append(cells[2].text.strip())
    last_eruption.append(cells[3].text.strip())
    coordinates.append(cells[4].text.strip().split("/")[0])
    island.append('Sunda Strait and Java')

rows = tables[2].find_all('tr')    
for row in rows[1:]:
    cells = row.find_all('td')
                
    name.append(cells[0].text.strip())
    shape.append(cells[2].text.strip())
    elevation.append(cells[3].text.strip())
    last_eruption.append(cells[4].text.strip())
    coordinates.append(cells[5].text.strip().split("/")[0])
    island.append(cells[1].text.strip())

rows = tables[3].find_all('tr')
for row in rows[1:]:
    cells = row.find_all('td')
                
    name.append(cells[0].text.strip())
    shape.append(cells[1].text.strip())
    elevation.append(cells[2].text.strip())
    last_eruption.append(cells[3].text.strip())
    coordinates.append(cells[4].text.strip().split("/")[0])
    island.append('Banda Sea')
        
rows = tables[4].find_all('tr')
for row in rows[1:]:
    cells = row.find_all('td')
                
    name.append(cells[0].text.strip())
    shape.append(cells[1].text.strip())
    elevation.append(cells[2].text.strip())
    last_eruption.append(cells[3].text.strip())
    coordinates.append(cells[4].text.strip().split("/")[0])
    island.append('Sulawesi and Sangihe Islands')
        

rows = tables[5].find_all('tr')
for row in rows[1:]:
    cells = row.find_all('td')
                
    name.append(cells[0].text.strip())
    shape.append(cells[1].text.strip())
    elevation.append(cells[2].text.strip())
    last_eruption.append(cells[3].text.strip())
    coordinates.append(cells[4].text.strip().split("/")[0])
    island.append('Halmahera')
                
data_volcano = {'Name' : name, 'Shape' : shape, 'Elevation' : elevation, 'Last Eruption' : last_eruption,
                'Coordinates' : coordinates, 'Island' : island}
df_volcano = pd.DataFrame(data_volcano)
```

```python
df_volcano.shape
df_volcano
```

```python
# The last table depict the major eruption that happened in the country.

eruption_date = []
volcano = []
cessation_date = []
vei = []
characteristics = []
tsunami = []
tephra_volume = []
fatality = []

row_erupt = tables[6].find_all('tr')
for row in row_erupt[1:]:
    cells_erupt = row.find_all('td')
            
    eruption_date.append(cells_erupt[0].text.strip())
    volcano.append(cells_erupt[1].text.strip())
    cessation_date.append(cells_erupt[2].text.strip())
    vei.append(cells_erupt[3].text.strip())
    characteristics.append(cells_erupt[4].text.strip())
    tsunami.append(cells_erupt[5].text.strip())
    tephra_volume.append(cells_erupt[6].text.strip())
    fatality.append(cells_erupt[7].text.strip())
    
data_eruption = {'Eruption Date' : eruption_date, 'Volcano' : volcano, 'Cesssation Date' : cessation_date, 'VEI' : vei, 
        'Characteristic': characteristics, 'Tsunami' : tsunami, 'Tephra Volume': tephra_volume, 'Fatality': fatality}       
df_eruption = pd.DataFrame(data_eruption)
```

```python
df_eruption.shape
df_eruption
```

```python
# We are going to save the dataset on two separate CSV files

df_volcano.to_csv("Indonesia_volcano.csv", sep=',', encoding='utf-8-sig')
df_eruption.to_csv("Indonesia_major_eruption.csv", sep=',', encoding='utf-8-sig')
```
